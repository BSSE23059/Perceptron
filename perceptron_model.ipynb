```python
import numpy as np
from sklearn.datasets import make_classification
X, y = make_classification(n_samples=100, n_features=2, n_redundant=0, n_informative=1,
                           n_classes=2, n_clusters_per_class=1, random_state=41, hypercube=False, class_sep=10)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 6))
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)
plt.show()
```

```python
def step(x):
    return 1 if x>0 else 0
def perceptron(x_, y_):
    x_ = np.insert(x_, 0, 1, axis=1)
    weights = np.zeros(x_.shape[1])
    lr = 0.1
    for i in range(1000):
        j = np.random.randint(0, x_.shape[0])
        y_hat = step(np.dot(x_[j], weights))
        weights = weights + lr * (y_[j] - y_hat) * x_[j]
    return weights[0], weights[1:]
```

```python
intercept_, coefficients_ = perceptron(X, y)
```

```python
m = -(coefficients_[0]/coefficients_[1])
b  = -(intercept_ / coefficients_[1])
```

```python
x_input = np.linspace(-3, 3, 100)
y_input = m * x_input + b
```

```python
plt.figure(figsize=(10, 6))
plt.plot(x_input, y_input, color='red', linewidth=3)
plt.scatter(X[:,0], X[:,1], c=y, cmap='winter', s=100)
plt.ylim(-3, 3)
plt.show()
```
